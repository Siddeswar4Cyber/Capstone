## Import Libraries

## Load All CSV Files

```python
import os
import glob

path = 'C:/Users/nimma/Downloads/IDS-PROJECT/dataset'

csv_files = glob.glob(os.path.join(path,"*.csv"))

print(csv_files)
```

## Merge All Files

```python
df = pd.concat(
    (pd.read_csv(file,low_memory=False) for file in csv_files),
    ignore_index = True
)
```
