#### Import Libraries

```python
import joblib as jl
import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

import matplotlib.pyplot as plt
```

#### Load Dataset

```python
X_train, X_val, X_test, y_train, y_val, y_test = jl.load(
    '/home/kali/Documents/IDS-Project/models/preprocessed_data.pkl'
)
```

#### Load Best Models

```python
rf = jl.load('/home/kali/Documents/IDS-Project/models/random_forest_best.pkl')
xgb = jl.load('/home/kali/Documents/IDS-Project/models/xgboost_best.pkl')
```

#### Create Evaluation Function

```python
def evaluate(model, X_test, y_test):
    pred = model.predict(X_test)

    return {
        "Accuracy": accuracy_score(y_test,pred),
        "Precision": precision_score(y_test, pred, average="weighted"),
        "Recall": recall_score(y_test, pred, average="weighted"),
        "F1": f1_score(y_test, pred, average="weighted"),
        "Prediction": pred,
    }
```

#### Evaluate Random Forest

```python
rf_result = evaluate(rf,X_test,y_test)

print(rf_result)
```

```python
xgb_result = evaluate(xgb, X_test, y_test)
print(xgb_result)
```

#### Classification Report

```python
print("Random Forest")
print(classification_report(
    y_test,
    rf_result["Prediction"]
))

print("XGBoost")
print(classification_report(
    y_test,
    xgb_result["Prediction"]
))
```

#### Confusion Matrix

```python
cm = confusion_matrix(
    y_test,
    rf_result['Prediction']
)
disp = ConfusionMatrixDisplay(cm)

disp.plot(
    xticks_rotation=90,
    cmap="Blues"
)
plt.show()
```
