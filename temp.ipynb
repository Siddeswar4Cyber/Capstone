# Step 4: Data Cleaning

## 4.1 Remove Duplicate Rows

```python
print("Before:", df.shape)

duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("After:", df.shape)
```

## 4.2 Clean Column Names

```python
df.columns = df.columns.str.strip()
```

```python
print(df.columns.tolist())
```

## 4.3 Replace Infinite Values

```python
numeric_df = df.select_dtypes(include=np.number)

print(np.isinf(numeric_df).sum().sort_values(ascending=False))
```

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
print(np.isinf(df.select_dtypes(include=np.number)).sum().sum())
```

## 4.4 Handle Missing Values

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print(missing)
```

```python
df.dropna(inplace=True)
```

```python
print(df.isnull().sum().sum())
```

## 4.5 Remove Constant Features

```python
constant_columns = []

for col in df.columns:
    if df[col].nunique() == 1:
        constant_columns.append(col)

print(constant_columns)
```

```python
df.drop(columns=constant_columns, inplace=True)
```

## 4.6 Check Data Types

```python
df.info()
```

## 4.7 Standardize Labels

```python
df["Label"] = df["Label"].str.strip()
```

```python
print(df["Label"].unique())
```

## 4.8 Verify Class Distribution

```python
df["Label"].value_counts()
```

```python
(df["Label"].value_counts(normalize=True) * 100).round(2)
```

## 4.9 Final Dataset Check

```python
print(df.shape)

print(df.isnull().sum().sum())

print(df.duplicated().sum())

print(np.isinf(df.select_dtypes(include=np.number)).sum().sum())
```
